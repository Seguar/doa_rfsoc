```python
def get_custom_data_xm655(self):
    """Data from 4 (I + Q) channels xm655, int16"""
    channels = 8  # Assuming there are 4 channels, but channels variable is set to 8
    # Initialize buffer with zeros
    buffer = np.zeros((2, len(self.adc_capture0)), dtype=np.int16)

    # Trigger ADC capture if needed
    self.trigger_capture()

    # Copy ADC capture data into the buffer
    buffer[0] = np.copy(self.adc_capture0[:len(buffer[0])])
    buffer[1] = np.copy(self.adc_capture1[:len(buffer[1])])

    final_buffer = np.zeros((len(self.adc_capture0)), dtype=np.int16)

    for i in range(0, len(self.adc_capture0), channels):
        final_buffer[0+i] = buffer[0][0+i]
        final_buffer[1+i] = buffer[0][1+i]
        final_buffer[2+i] = buffer[0][4+i]
        final_buffer[3+i] = buffer[0][5+i]
        final_buffer[4+i] = buffer[1][0+i]
        final_buffer[5+i] = buffer[1][1+i]
        final_buffer[6+i] = buffer[1][4+i]
        final_buffer[7+i] = buffer[1][5+i]

    # Ensure final buffer is trimmed to data_size
    custom_data = final_buffer[:self.data_size]

    return custom_data
```

```python
def get_custom_data_xm655(self):
    """Data from 4 (I + Q) channels xm655, int16"""
    channels = 8  # Assuming there are 4 channels, but the variable is set to 8

    # Trigger ADC capture if needed
    self.trigger_capture()

    # Buffer the relevant channels directly
    buffer = np.zeros((2, len(self.adc_capture0)), dtype=np.int16)
    buffer[0] = self.adc_capture0
    buffer[1] = self.adc_capture1

    # Prepare the final buffer using advanced indexing and slicing
    final_buffer = np.zeros(len(self.adc_capture0), dtype=np.int16)
    
    # Use vectorized operations instead of a loop
    final_buffer[0::channels] = buffer[0][0::channels]
    final_buffer[1::channels] = buffer[0][1::channels]
    final_buffer[2::channels] = buffer[0][4::channels]
    final_buffer[3::channels] = buffer[0][5::channels]
    final_buffer[4::channels] = buffer[1][0::channels]
    final_buffer[5::channels] = buffer[1][1::channels]
    final_buffer[6::channels] = buffer[1][4::channels]
    final_buffer[7::channels] = buffer[1][5::channels]

    # Trim final buffer to data_size
    custom_data = final_buffer[:self.data_size]

    return custom_data
```

```python
def handle_client_connection(self, client_socket):
    while True:
        # Receive data from client until terminator is found
        data = b''
        while True:
            chunk = client_socket.recv(1)  # Receive one byte at a time
            if chunk == b'\r' or chunk == b'\n' or chunk == b'':  # ASCII CR or LF character
                break
            data += chunk

        if not data:
            break

        message = data.decode('utf-8')
        return message
```

```python
def handle_commands(self, commandList):
        for command_str in commandList:
            command, var = command_str.split()
            var = var.split('/')            
            var = list(map(int, var))
            match command:
                case "fc":
                    self.centre_freq = [var[0], var[0], var[0], var[0]]
                    self.nyquist_zone = [var[1], var[1], var[1], var[1]] 
                    self.d_centre_freq = [var[2], var[2], var[2], var[2]]
                    self.d_nyquist_zone = [var[3], var[3], var[3], var[3]]
                    self.configure_dacs()
                    self.configure_adcs()
                    # self.sync_tiles()
                case "cal": # Only for ADCs
                    self.channels = var[0]
                    # self.calibrate()
                    self.calibrate_xm655()
                case "phase":
                    # self.phases = var
                    self.phases[0] = var[0]
                    self.phases[2] = var[1]
                    self.phases[4] = var[2]
                    self.phases[6] = var[3]
                    self.configure_adcs()
                    # self.sync_tiles()
                case "dphase":
                    # self.d_phases = var
                    self.d_phases[0] = var[0]/100
                    self.d_phases[2] = var[1]/100
                    self.d_phases[4] = var[2]/100
                    self.d_phases[6] = var[3]/100
                    
                    self.d_phases[8] = var[3]/100
                    self.d_phases[10] = var[2]/100
                    self.d_phases[12] = var[1]/100
                    self.d_phases[14] = var[0]/100
                    # self.configure_dac(8, self.d_nyquist_zone[1], self.d_centre_freq[1], self.event_src, self.d_phases[8], self.d_gain[8])
                    # self.configure_dac(10, self.d_nyquist_zone[1], self.d_centre_freq[1], self.event_src, self.d_phases[10], self.d_gain[10])
                    # self.configure_dac(12, self.d_nyquist_zone[1], self.d_centre_freq[1], self.event_src, self.d_phases[12], self.d_gain[12])
                    # self.configure_dac(14, self.d_nyquist_zone[1], self.d_centre_freq[1], self.event_src, self.d_phases[14], self.d_gain[14])
                    
                    self.dac[0].MixerSettings['PhaseOffset'] = self.d_phases[0]
                    self.dac[2].MixerSettings['PhaseOffset'] = self.d_phases[2]
                    self.dac[4].MixerSettings['PhaseOffset'] = self.d_phases[4]
                    self.dac[6].MixerSettings['PhaseOffset'] = self.d_phases[6]
                    self.dac[8].MixerSettings['PhaseOffset'] = self.d_phases[8]
                    self.dac[10].MixerSettings['PhaseOffset'] = self.d_phases[10]
                    self.dac[12].MixerSettings['PhaseOffset'] = self.d_phases[12]
                    self.dac[14].MixerSettings['PhaseOffset'] = self.d_phases[14]
                    # for i in range(len(self.dac)):
                    #     self.dac[i].MixerSettings['PhaseOffset'] = self.d_phases[i]

                    # self.configure_dacs()
                    # self.sync_tiles()
                case "again":
                    # self.phases = var
                    self.a_gain[0] = var[0]/100
                    self.a_gain[2] = var[1]/100
                    self.a_gain[4] = var[2]/100
                    self.a_gain[6] = var[3]/100
                    self.configure_adcs()
                case "dgain":
                    # self.d_phases = var
                    self.d_gain[0] = var[0]/100
                    self.d_gain[2] = var[1]/100
                    self.d_gain[4] = var[2]/100
                    self.d_gain[6] = var[3]/100
                    
                    self.d_gain[8] = var[3]/100
                    self.d_gain[10] = var[2]/100
                    self.d_gain[12] = var[1]/100
                    self.d_gain[14] = var[0]/100
                    self.configure_dacs()
                case "dataChan":
                    self.data_size = var[0]

                case "dataStream":              
                    self.dataStream = var[0]
                case "da":
                    self.da = var[0]
                    match self.da:
                        case 2: # DACs = ADCs sync
                            self.dacs_off()
                            self.trig_cap.off()
                            self.trig_cap.on() 
                            self.trig_cap.off()
                        case 1: # DACs = DACs sync
                            self.dacs_off()
                            self.trig_cap.off()
                            self.trig_cap.on() 
                            self.dacs_control()
                            self.trig_cap.off()                            
                        case 0: # DACs =/= ADCs without sync
                            self.dacs_off()
                            self.dacs_control()
                            
                case "source":
                    self.source = var[0]

                    DAC_SR = var[1]*1E9  # Sample rate of DACs (10GSPS)
                    print(DAC_SR)
                    Fc = var[2]*1E6 # Center frequency of waveform
                    print(Fc)
                    Fe = var[3]*1E6 # Maximum frequency of chirp at end of record
                    print(Fe)
                    DAC_Amplitude = float(var[4]) # 14bit DAC +16383/-16384     
                    print(DAC_Amplitude)
                    X_axis = (1/DAC_SR) * np.arange(0, self.dac0_player.shape[0]//2)
                    print('Params loaded')
                    self.signal = np.zeros(self.dac0_player.shape[0], dtype=np.int16)
                    match self.source:
                        case 4: # File
                            mat = scipy.io.loadmat('data.mat')
                            self.signal = mat.get('sigInt16')
                            self.signal = self.signal[0]
                        case 3: # Chirp
                            self.signal[0::2] = np.int16(DAC_Amplitude * chirp(np.arange(0, self.dac0_player.shape[0]/2)/DAC_SR,
                                      f0=Fc, f1=Fe, t1=(self.dac0_player.shape[0]/DAC_SR), 
                                      phi=0.0, method='linear'))
                            self.signal[1::2] = np.int16(DAC_Amplitude * chirp(np.arange(0, self.dac0_player.shape[0]/2)/DAC_SR,
                                      f0=Fc, f1=Fe, t1=(self.dac0_player.shape[0]/DAC_SR), 
                                      phi=90.0, method='linear'))
                        case 2: # SAW
                            self.signal[0::2] = np.int16(DAC_Amplitude * sawtooth(2 * np.pi * Fc * X_axis))
                            self.signal[1::2] = np.int16(DAC_Amplitude * sawtooth(2 * np.pi * Fc * X_axis + np.pi/2))
                        case 1: # CW
                            self.signal[0::2] = np.int16(DAC_Amplitude * np.sin(2*np.pi*Fc*X_axis))
                            self.signal[1::2] = np.int16(DAC_Amplitude * np.cos(2*np.pi*Fc*X_axis))
                            
                        case 0: # Zeros
                            self.signal = np.int16(np.zeros(len(self.dac0_player)))
                    print('Signal ready')
                    self.dac_data_mem_write(self.signal, self.dac0_player)
                    self.dac_data_mem_write(self.signal, self.dac1_player)
                    self.dac_data_mem_write(self.signal, self.dac2_player)
                    self.dac_data_mem_write(self.signal, self.dac3_player)
                    print('Signal loaded')
                case "dac0":    
                    self.dac_signal[0] = var
            
                    self.dac_data_mem_write(self.dac_signal[0], self.dac0_player)
            
                case "dac1":
                    self.dac_signal[2] = var
            
                    self.dac_data_mem_write(self.dac_signal[2], self.dac1_player)
            
                case "dacPow":
                    self.dacPower = var
                    
                case "sync":
                    result = None
                    while result is None:
                        try:
                            # connect
                            self.sync_tiles()
                            result = True
                            print('Success')
                        except:
                            time.sleep(0.1)
                            print('Error')
                            pass
```
